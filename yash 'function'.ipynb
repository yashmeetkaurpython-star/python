```python
#factorial program
def fact(n):
    if n==0:
        return 1
    elif n==1:
        return 1
    else:
        return n*fact(n-1)
n=int(input("enter the value of n:="))
print(fact(n))   
```

```python
def fib(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
n=int(input("enter any number"))
for i in range(n):
    print(fib(i))
```
