```python
def info(name,/):
    print(f"Hello And Welcome to All my name is {name}")
info("Yash")
info("Dilpreet")
info("Aman")          
info("Saman")          
```

```python
def info(*,name):
    print(f"Hello And Welcome to All my name is {name}")
info(name="Yash")
info(name="Dilpreet")
info(name="Aman")          
info(name="Saman")          
```

```python
def info(name,/,*,fname):
    print(f"my name is {name} and my parent name is {fname}")
info("Yash",fname="Rajinder Singh")    
```

```python
def info(*name):
    print("hello",name[0])
info("Yash","Dilpreet","Aman","Saman")    
```

```python
def info(**name):
    print("Hello my name is:",name)
info(name="Yash")    
```
